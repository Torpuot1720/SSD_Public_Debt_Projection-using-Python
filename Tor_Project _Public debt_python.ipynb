```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
%matplotlib inline 
import os

df = pd.read_csv('SSD_Public_Debt_Principal Repayment.csv')

df.info() 

# Standardize column names and fix the typo in the debt column before using it
df.columns = df.columns.map(lambda c: str(c).strip())
df.columns = df.columns.map(lambda c: " ".join(c.split()))

# Try the expected name first, then the actual typo present in the CSV
candidates = [
    "Outstanding Debt - end FY2023/2024",
    "Outstandind Debt - end FY2023/2024",
]

debt_col = next((col for col in df.columns if col in candidates), None)

if debt_col is None:
    # Case-insensitive/fuzzy fallback
    normalized = {str(c).lower(): c for c in df.columns}
    debt_col = (
        normalized.get("outstanding debt - end fy2023/2024")
        or normalized.get("outstandind debt - end fy2023/2024")
        or next(
            (col for col in df.columns if "debt" in str(col).lower() and "fy2023/2024" in str(col).lower()),
            None
        )
    )

if debt_col is None:
    raise KeyError("Outstanding Debt - end FY2023/2024 column not found")

if debt_col != "Outstanding Debt - end FY2023/2024":
    df = df.rename(columns={debt_col: "Outstanding Debt - end FY2023/2024"})

# Convert to numeric safely
df["Outstanding Debt - end FY2023/2024"] = pd.to_numeric(
    df["Outstanding Debt - end FY2023/2024"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace(" ", "", regex=False),
    errors="coerce"
)
# Convert outstanding debt to numeric
# Fix the typo in the debt column name if present
debt_col = next(
    (col for col in df.columns if "Debt" in col and "FY2023/2024" in col),
    None
)

if debt_col is None:
    raise KeyError("Outstanding Debt - end FY2023/2024 column not found")

if debt_col != "Outstanding Debt - end FY2023/2024":
    df = df.rename(columns={debt_col: "Outstanding Debt - end FY2023/2024"})

# Convert to numeric and remove thousands separators if any
df["Outstanding Debt - end FY2023/2024"] = pd.to_numeric(
    df["Outstanding Debt - end FY2023/2024"]
    .astype(str)
    .str.replace(",", "", regex=False),
    errors="coerce"
)

# Group debt by creditor and sort from smallest to largest
# Normalize column names to avoid hidden spaces or formatting issues
df.columns = df.columns.map(lambda c: str(c).strip())
df.columns = df.columns.map(lambda c: " ".join(c.split()))

# Find the actual creditor-name column even if the header has extra spaces or minor typos
creditor_col = next(
    (col for col in df.columns if "creditor" in col.lower() and "name" in col.lower()),
    None
)

if creditor_col is None:
    raise KeyError("Creditor Name column not found")

if creditor_col != "Creditor Name":
    df = df.rename(columns={creditor_col: "Creditor Name"})

# Find the debt column even if the header has spacing/typo differences
debt_col = next(
    (col for col in df.columns if "outstanding debt" in col.lower() and "fy2023/2024" in col.lower()),
    None
)

if debt_col is None:
    raise KeyError("Outstanding Debt - end FY2023/2024 column not found")

if debt_col != "Outstanding Debt - end FY2023/2024":
    df = df.rename(columns={debt_col: "Outstanding Debt - end FY2023/2024"})

# Clean numeric values
df["Outstanding Debt - end FY2023/2024"] = pd.to_numeric(
    df["Outstanding Debt - end FY2023/2024"]
    .astype(str)
    .str.replace(",", "", regex=False),
    errors="coerce"
)

# Aggregate debt by creditor
debt_by_creditor = (
    df.groupby("Creditor Name", dropna=False)["Outstanding Debt - end FY2023/2024"]
    .sum()
    .sort_values(ascending=True)
)

# Create horizontal bar chart
plt.figure(figsize=(12, 7))

debt_by_creditor.plot(
    kind="barh",
    color="steelblue",
    edgecolor="black"
)

plt.title(
    "South Sudan: Outstanding Public Debt by Creditor\nFY 2023/2024",
    fontsize=14
)
plt.xlabel("Outstanding Debt (recorded currency units)", fontsize=11)
plt.ylabel("Creditor Name", fontsize=11)

plt.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()













```
